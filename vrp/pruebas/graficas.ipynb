```python
import plotly.express as px
import pandas as pd
import os

# Create a directory to save the plots
output_dir = 'plotly_frequency_plots'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

def plot_minimum_frequency_by_class(df):
    min_class_data = []

    instances = df['Nombre de la Instancia VRP'].unique()
    for instance in instances:
        subset = df[df['Nombre de la Instancia VRP'] == instance]
        for combination in subset['Combinacion'].unique():
            comb_data = subset[subset['Combinacion'] == combination]
            min_cost = comb_data['Best Cost'].min()
            min_class_data.append({
                'Instancia': instance,
                'Clase': combination,
                'Min Cost': min_cost
            })

    min_class_df = pd.DataFrame(min_class_data)

    # Find the minimum cost for each instance
    min_class_df = min_class_df.loc[min_class_df.groupby('Instancia')['Min Cost'].idxmin()]

    frequency_df = min_class_df.groupby(['Instancia', 'Clase']).size().reset_index(name='Frecuencia')

    fig = px.bar(frequency_df, x='Instancia', y='Frecuencia', color='Clase', barmode='group',
                 title='Frecuencia de la Clase que Obtiene el Mínimo Cost por Instancia',
                 labels={'Frecuencia': 'Frecuencia de Mínimos', 'Instancia': 'Nombre de la Instancia VRP', 'Clase': 'Clase'})

    fig.write_image(os.path.join(output_dir, 'frequency_plot.png'))
    fig.show()

# Call the function with the DataFrame to save the plot
plot_minimum_frequency_by_class(df)
```

```python
import plotly.graph_objects as go
import os

# Create a directory to save the plots
output_dir = 'plotly_error_plots'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

def plot_error_by_iteration_plotly(df):
    instances = df['Nombre de la Instancia VRP'].unique()
    for instance in instances:
        subset = df[df['Nombre de la Instancia VRP'] == instance]
        solution = float(subset['Solucion'].values[0])
        fig = go.Figure()

        colors = ['blue', 'green', 'orange']  # Colors for different combinations
        color_index = 0

        for combination in subset['Combinacion'].unique():
            comb_data = subset[subset['Combinacion'] == combination]
            iteration_labels = ['Cost_i1', 'Cost_i2', 'Cost_i3', 'Cost_i4', 'Cost_i5', 'Cost_i6']
            color = colors[color_index % len(colors)]
            color_index += 1

            for i, iteration in enumerate(iteration_labels):
                iteration_costs = comb_data[iteration].values
                min_error = min(abs(iteration_costs - solution))
                min_error_index = list(abs(iteration_costs - solution)).index(min_error)
                min_error_cost = iteration_costs[min_error_index]
                fig.add_trace(go.Scatter(
                    x=[f'{combination}_{iteration}'] * len(iteration_costs),
                    y=iteration_costs,
                    mode='markers+lines',
                    line=dict(color=color, width=2),
                    marker=dict(color=color, size=8),
                    name=f"{combination} - {iteration}"
                ))
                fig.add_trace(go.Scatter(
                    x=[f'{combination}_{iteration}'],
                    y=[solution],
                    mode='markers',
                    marker=dict(color='red', size=12),
                    name=f"Solution - {iteration}"
                ))
                fig.add_trace(go.Scatter(
                    x=[f'{combination}_{iteration}'],
                    y=[min_error_cost],
                    mode='markers',
                    marker=dict(color='purple', size=12),
                    name=f"Min Error - {combination} - {iteration}"
                ))

        fig.update_layout(
            title=f'Error by Solution for Each Iteration - {instance}',
            xaxis_title='Iterations',
            yaxis_title='Cost',
            legend_title='Combinations',
            template='plotly_white'
        )

        fig.write_image(os.path.join(output_dir, f'errorplot_{instance}.png'))
        fig.show()

# Call the function with the DataFrame to save the plots
plot_error_by_iteration_plotly(df)
```
